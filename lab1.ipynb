```python
import numpy as np
import pandas as pd
print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)
```

```python
marks=np.array([88,36,41,92,33])
```

```python
print("Number of Dimensions:", marks.ndim)
print("Shape:", marks.shape)
print("Number of Dimensions:", marks.ndim)
```

```python
marks=np.array([88,36,41,92,33])
print("Original:", marks)
print("Add 5:", marks+5)
print("Multiply by 2:", marks*2)
print("Divide by 2:", marks/2)
print("Square:", marks**2)
```

```python
print("Sum:", marks.sum())
print("Mean:", marks.mean())
print("Max:", marks.max())
print("Min:", marks.min())
```

```python
print("First mark:", marks[0])
print("Second mark:", marks[1])
print("Last mark:", marks[-1])
```

```python
print("Index 1 to 3:", marks[1:4])
print("First 3:", marks[:3])
print("From Index 2:", marks[2:])
print("Every second element:", marks[::2])
```

```python
studend_marks = np.array([
    [36, 41, 92],
    [88, 36, 41],
    [90, 95, 89]
])
print(studend_marks)
print("shape:", studend_marks.shape)
```

```python
print("Mean:", studend_marks.mean())
print("Column-wise mean:", studend_marks.mean(axis=0))
print("Row-wise mean:", studend_marks.mean(axis=1))
```
